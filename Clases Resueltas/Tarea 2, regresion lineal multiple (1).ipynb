```python
import pandas as pd
from sklearn import linear_model
from sklearn.metrics import r2_score 
```

```python
df = pd.read_csv('../Data/day.csv')
```

```python
y = df['cnt']

X = df.copy()
del X['cnt']
del X['casual']
del X['registered']
del X['instant']
del X['dteday']
del X['mnth']
del X['yr']
```

```python
categoricas = [
    'season',
    'holiday',
    'weekday',
    'workingday',
    'weathersit'
]
```

```python
X = pd.get_dummies(X, columns=categoricas)
```

```python
model = linear_model.LinearRegression().fit(X,y)
```

```python
predicciones_todo = model.predict(X)
```

```python
r2_score(y_pred=predicciones_todo, y_true=y)
```

El modelo obtuvo un \(R^2 = 0.5642\), por lo que explica aproximadamente el 56.42% de la variación en la cantidad de bicicletas rentadas.
