```python
import pandas as pd
from sklearn import linear_model
from sklearn.metrics import r2_score
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures


df = pd.read_csv('../Data/housing.csv')
target = 'median_house_value'
df = pd.get_dummies(df)
```

```python
df['feature'] = df['households'] / df['population']
```

```python
# Regresion lineal feature
X = df[['feature']]
y = df['median_house_value']
model= linear_model.LinearRegression()
model.fit(X,y)
predicciones_lineal=model.predict(X)
r2_score(y_pred=predicciones_lineal, y_true=y)
```

```python
# Regresion polinomial grado 2
degree = 4
# Define poly features
poly_feature = PolynomialFeatures(degree=degree)
# Transformalos
X_poly=poly_feature.fit_transform(X)
# Ajustar el modelo de regresión lineal
model_poly2=linear_model.LinearRegression()
model_poly2.fit(X_poly,y)
# saca el r2
predicciones_poly2=model_poly2.predict(X_poly)
r2_score(y_pred=predicciones_poly2, y_true=y)
```

### Cross validation 

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
```

```python
# Crea el pipeline
pipeline=Pipeline([
  ('poly',PolynomialFeatures()),
  ('model',linear_model.LinearRegression())
])

params={
    'poly__degree':[1,2,3,4]
}

# Grados a probar

# Cross-validation
grid=GridSearchCV(estimator=pipeline, param_grid=params, scoring='r2', cv=5)
grid.fit(X, y)
```

### Crea una regresión polinomial con todos y has cross validación 

```python
df = pd.read_csv('../Data/housing.csv')
target = 'median_house_value'
df = pd.get_dummies(df)
```

```python
df['density'] = df['households'] / df['population']
```

```python
X = df.copy()
X = X.drop(target, axis=1)
y = df[target]
X['total_bedrooms'] = X.total_bedrooms.fillna(df.total_bedrooms.mean())

del X['households']
del X['population']

X = pd.get_dummies(X)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=1)
```
